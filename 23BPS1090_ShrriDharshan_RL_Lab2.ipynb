### Aim
To implement and compare three Multi-Armed Bandit algorithms—Epsilon-Greedy, Upper Confidence Bound (UCB), and Thompson Sampling—for solving the exploration-exploitation problem and evaluate their performance based on cumulative rewards.

```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from bandits.environment import MultiArmedBanditEnv
from bandits.agents import EpsilonGreedyBandit, UCBandit, ThompsonBandit
```

### 1. Environment Setup & True Arm Means
We define a 10-armed Gaussian testbed where each arm generates rewards from a normal distribution with fixed true mean.

```
env = MultiArmedBanditEnv(num_actions=10, seed=42)
print("True reward means per arm:", np.round(env.arm_means, 2))
print(f"Optimal arm: #{env.optimal_arm} with mean reward = {env.arm_means[env.optimal_arm]:.2f}")
```

### 2. Simulation Benchmark
We simulate each algorithm for 1,000 steps and log both cumulative rewards and cumulative regret.

```
num_steps = 1000
eps = 0.1

eg_agent = EpsilonGreedyBandit(10, eps)
ucb_agent = UCBandit(10)
ts_agent = ThompsonBandit(10)

eg_rewards, ucb_rewards, ts_rewards = [], [], []
eg_tot, ucb_tot, ts_tot = 0, 0, 0

for step in range(num_steps):
    # Epsilon Greedy
    a_eg = eg_agent.choose_action()
    r_eg = env.step(a_eg)
    eg_agent.update(a_eg, r_eg)
    eg_tot += r_eg
    eg_rewards.append(eg_tot)

    # UCB
    a_ucb = ucb_agent.choose_action()
    r_ucb = env.step(a_ucb)
    ucb_agent.update(a_ucb, r_ucb)
    ucb_tot += r_ucb
    ucb_rewards.append(ucb_tot)

    # Thompson Sampling
    a_ts = ts_agent.choose_action()
    r_ts = env.step(a_ts)
    ts_agent.update(a_ts, r_ts)
    ts_tot += r_ts
    ts_rewards.append(ts_tot)

print(f"Epsilon-Greedy Total Reward: {eg_tot:.2f}")
print(f"UCB Total Reward:            {ucb_tot:.2f}")
print(f"Thompson Sampling Reward:    {ts_tot:.2f}")
```

### 3. Cumulative Reward Trajectory Visualization

```
plt.figure(figsize=(10, 5))
plt.plot(eg_rewards, label='Epsilon-Greedy', color='#2b5c8f', lw=2)
plt.plot(ucb_rewards, label='UCB', color='#d9534f', lw=2)
plt.plot(ts_rewards, label='Thompson Sampling', color='#5cb85c', lw=2)
plt.title('Multi-Armed Bandit: Cumulative Reward over Steps', fontsize=13, fontweight='bold')
plt.xlabel('Steps', fontsize=11)
plt.ylabel('Cumulative Reward', fontsize=11)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### 4. Conclusion & Findings
The Multi-Armed Bandit algorithms Epsilon-Greedy, Upper Confidence Bound (UCB), and Thompson Sampling were successfully implemented and evaluated. Thompson Sampling and UCB balance exploration and exploitation more efficiently than simple $\epsilon$-Greedy by actively directing exploration toward uncertain or promising arms.
